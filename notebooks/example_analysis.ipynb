# 气候数据分析示例
本 Notebook 演示如何使用 Climate Agent 工具集进行探索性分析

```python
import sys
import os
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from src.tools.analysis.statistics import compute_trend
from src.tools.visualization.plotter import plot_timeseries
from src.tools.utils import load_data_file, get_output_path
```

```python
# 生成模拟数据
years = np.arange(2010, 2025)
temperature = 15 + 0.03 * (years - 2010) + np.random.randn(len(years)) * 0.2

# 计算趋势
result = compute_trend(years, temperature)
print(f"趋势斜率: {result['slope']:.4f} °C/年")
print(f"R²: {result['r_squared']:.4f}")
print(f"p值: {result['p_value']:.4f}")
```

```python
# 可视化
plot_timeseries(
    years,
    temperature,
    title="全球平均温度变化",
    xlabel="年份",
    ylabel="温度 (°C)",
    save_path="../outputs/figures/notebook_trend.png",
    show=True
)
```

## 使用记忆模块存储分析结果
